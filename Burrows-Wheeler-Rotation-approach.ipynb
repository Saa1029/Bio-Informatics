```python
def bwt_transform(text):
    """
    Compute the Burrows-Wheeler Transform using the rotation method.
    """

    if "$" not in text:
        text = text + "$"

    if text.count("$") != 1:
        raise ValueError("Input must contain exactly one '$' end marker.")

    rotations = []
    n = len(text)

    for i in range(n):
        rotation = text[i:] + text[:i]
        rotations.append(rotation)

    rotations.sort()

    bwt = ""

    for rotation in rotations:
        bwt += rotation[-1]

    return bwt


def bwt_reconstruct(bwt):
    """
    Reconstruct the original string from BWT using the rotation matrix method.
    """

    if bwt.count("$") != 1:
        raise ValueError("BWT string must contain exactly one '$' end marker.")

    n = len(bwt)
    table = [""] * n

    for _ in range(n):
        table = sorted([bwt[i] + table[i] for i in range(n)])

    for row in table:
        if row.endswith("$"):
            return row

    raise ValueError("Could not reconstruct original string.")


def bwt_reconstruct_steps(bwt):
    """
    Reconstruct the original string from BWT and print each step.
    """

    if bwt.count("$") != 1:
        raise ValueError("BWT string must contain exactly one '$' end marker.")

    n = len(bwt)
    table = [""] * n

    print("Initial empty table:")
    for row in table:
        print(repr(row))
    print()

    for step in range(1, n + 1):
        table = sorted([bwt[i] + table[i] for i in range(n)])

        print(f"After iteration {step}:")
        for row in table:
            print(row)
        print()

    for row in table:
        if row.endswith("$"):
            return row

    raise ValueError("Could not reconstruct original string.")


# -------------------------
# Examples
# -------------------------

examples = [
    "BANANA",
    "ctatatat",
    "appellee"
]

for text in examples:
    transformed = bwt_transform(text)
    reconstructed = bwt_reconstruct(transformed)

    print("Original:     ", text + "$")
    print("BWT:          ", transformed)
    print("Reconstructed:", reconstructed)
    print()
```
